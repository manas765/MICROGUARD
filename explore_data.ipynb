```python
import pandas as pd

df = pd.read_excel("MF Bank loan Dataset.xlsx")
df.head()
```

```python
print(df.shape)
df.info()
```

```python
df['loan_status'].value_counts()
```

```python
df['is_risky'] = df['loan_status'].apply(lambda x: 1 if x == 'COLLECTION' else 0)
df['is_risky'].value_counts()
```

```python
df['Mode of application'] = df['Mode of application'].fillna(df['Mode of application'].mode()[0])
df['past_due_days'] = df['past_due_days'].fillna(0)
df['Guarantor'] = df['Guarantor'].fillna('No')

df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
print(df[['Principal', 'terms', 'age', 'past_due_days']].describe())
```
